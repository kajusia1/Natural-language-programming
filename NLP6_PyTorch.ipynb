Zadanie3: Zdefiniuj dane uczące  X=np.array([1,2,3,4,5])  oraz  Y=np.array([−1,3,13,29,51]) . Zakładamy, że pomiędzy  X  i  Y  istnieje przybliżona zależność $Y=aX^2+bX+c$. Zbuduj model przewidujący wartość parametrów  a,b,c  w oparciu o  X  i  Y. Następnie wyznacz przewidywaną wartość dla  x=6 .

```python
X=np.array([1,2,3,4,5])
Y=np.array([-1,3,13,29,51])

a = 4.0
b = -6.0
c = 0.0

#Funkcja modelu
def forward(x):
    return a * (x**2) + b * x + c

#Loss fucntion
def loss(y, y_pred):
    return np.mean(((y_pred - y)**2))

#Gradient
def gradient(x, y, y_pred):
    return np.mean(2*(x**2)*(y_pred - y)), np.mean(2*x*(y_pred - y)), np.mean(2*(y_pred - y)) #pochodne (po funkcji kosztu)

print(f'Ile wynosi f(5): {forward(5):.3f}')
print(f'Ile wynosi f(6): {forward(6):.3f}')

#trenowanie
learning_rate = 0.004 #współczynnik uczenia 
n_iters = 1000 #liczba iteracji

for epoch in range(n_iters):
    y_pred = forward(X) #przewidywane wartości dla X
    l = loss(Y, y_pred) #wartość funkcji kosztu
    dl = gradient(X, Y, y_pred) #gradient
    a -= learning_rate * dl[0]    #aktualizujemy parametr a
    b -= learning_rate * dl[1]    ##aktualizujemy parametr b
    c -= learning_rate * dl[2]    ##aktualizujemy parametr c
    
    if epoch % 200 == 0: #wypritowanie parametów co 200 krok
        print(f'epoch {epoch+1}: a = {a:.3f}, b = {b:.3f}, c = {c:.3f}, loss = {l:.8f}')
  
print(f'Ile wynosi f(5) po trenowaniu: {forward(5):.3f}') #do ustalenia paramtrów, 
                                                        #powinno byc jak najbliżej 51
print(f'Ile wynosi f(6) po trenowaniu: {forward(6):.3f}')
```

Zbuduj model przewidujący wartość parametrów  a,b,c,d,e  w oparciu o  X  i  Y . Przetestuj trzy rodzaje parametrów, wyświetlając za każdym razem tylko 5 epok z modelu (w równych odstępach czasu). Skomentuj otrzymane wyniki

- parametry początkowe: a=0, b=0, c=0, d=0, e=0; learning_rate=0.01; n_iters = 10

Zadanie5. Utwórz dwa tensory jednowymiarowe  x=tensor([1,2,3,4,5,6])  oraz  y=tensor([7,8,3,−1.,−2,2]) . Następnie wyznacz nowy tensor wg przepisu  $z=x^3+2y$. Utwórz dwa nowe tensory (2D - o wymiarach 2 na 3 oraz 3D - o wymiarach 2,1,3 odpowiednio) poprzez zmianę wymiaru  z .

```python
import torch
x= torch.tensor([1,2,3,4,5,6])
y= torch.tensor([7,8,3,-1.,-2,2])

z = x**3 + 2*y
print(z)

z2 = z.view([2,-1])
print(z2)

z3 = z.view([2,1,3]) # 2 'bloki', każdy z 1 wierszem i 3 kolumnai.
print(z3)
```

<h4> Zadanie6: Wyjaśnij poniższe liniki kodu i otrzymany wynik (łącznie z uzyskanymi wartościami w x.grad)


```python
x = torch.tensor([-3.0,4,1], requires_grad=True) #tensor z opcją śledzenia 
z = x**3 #podnosimy kazdy elemet x do 3 potegi
print(z)
r = z.sum() #sumujemy elemety z z (64-26=38)
print(r)
r.backward() #obliczamy gradienty (3x**2)
print(x.grad) #printujemy gradienty  x

#tensor([27., 48.,  3.]) = [3*(-3)^2 3*4^2, 3*1^2] = [27, 48, 3]
```

```python
x = torch.tensor([-3.0,4,1], requires_grad=True) #tworzymy tensory z opcją śledzenia
y = torch.tensor([2.0,0,-1], requires_grad=True)
z = (x**2)*y #podnosimy każdy elem z x do 2 potegi i mnożymy przez odpowiadające elementy z y
r = z.sum() #sumujemy elemety z z (18-1=17)
print(r)
r.backward() #obliczamy gradienty 
print(x.grad) #wyswietlamy gradiety względem x (2xy)
#tensor([-12.,   0.,  -2.]) = [2*(-3)*2, 2*4*0, 2*1*(-1)] = [-12, 0, -2]
```
